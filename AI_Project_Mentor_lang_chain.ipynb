```python
from transformers import pipeline

print("Transformers is working!")
```

```python
generator = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct"
)
```

```python
response = generator(
    "Give me one unique AI project idea for an engineering student.",
    max_new_tokens=100
)

print(response[0]["generated_text"])
```

```python
from langchain_huggingface import HuggingFacePipeline

llm = HuggingFacePipeline(pipeline=generator)

print("LangChain connected to the AI model!")
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    input_variables=["idea"],
    template="""
You are an AI Project Mentor.

A student wants to build this project:

{idea}

Give the student:

1. Project Title
2. Problem Statement
3. Objectives
4. Recommended Technologies
5. Dataset
6. Development Steps
7. Future Improvements

Keep the explanation simple and practical.
"""
)

print("Prompt created!")
```

```python
chain = prompt | llm

print("AI Project Mentor chain created!")
```

```python
response = chain.invoke({
    "idea": "AI system that detects plant diseases from leaf images"
})

print(response)
```

```python
def project_mentor(idea):
    response = chain.invoke({
        "idea": idea
    })

    print("===== AI PROJECT MENTOR =====")
    print(response)
```

```python
project_mentor("AI system that detects plant diseases from leaf images")
```

```python
idea = input("Enter your project idea: ")

project_mentor(idea)
```

```python
from langchain_core.prompts import PromptTemplate

mentor_prompt = PromptTemplate(
    input_variables=["idea", "level", "technology"],
    template="""
You are an AI Project Mentor for engineering students.

Project Idea: {idea}
Student Level: {level}
Preferred Technology: {technology}

Create a practical project plan.

Include:
1. Project Title
2. Problem Statement
3. Objectives
4. Required Technologies
5. Dataset
6. System Architecture
7. Development Steps
8. Folder Structure
9. Testing Ideas
10. Future Improvements

Keep the explanation simple and suitable for the student's level.
"""
)

print("Mentor prompt created successfully!")
```

```python
mentor_chain = mentor_prompt | llm

print("AI Project Mentor chain created!")
```

```python
response = mentor_chain.invoke({
    "idea": idea,
    "level": level,
    "technology": technology
})

print("===== AI PROJECT MENTOR =====")
print(response)
```

```python
from langchain_core.prompts import PromptTemplate

idea_prompt = PromptTemplate(
    input_variables=["domain", "level"],
    template="""
You are an AI Project Mentor.

Generate 5 unique and practical Artificial Intelligence
and Machine Learning project ideas.

Domain: {domain}
Student Level: {level}

For each project, provide:
1. Project Title
2. Problem it solves
3. Main AI/ML technique
4. Suggested technology

Avoid very common projects such as:
- Basic chatbot
- Iris flower classification
- MNIST digit recognition
- Simple house price prediction

Make the ideas suitable for a student project.
"""
)

print("Project Idea Generator created!")
```

```python
idea_chain = idea_prompt | llm

print("Idea Generator connected to LangChain!")
```

```python
ideas = idea_chain.invoke({
    "domain": "Healthcare",
    "level": "Beginner"
})

print("===== PROJECT IDEAS =====")
print(ideas)
```

```python
domain = input("Enter your project domain: ")
level = input("Enter your skill level: ")

ideas = idea_chain.invoke({
    "domain": domain,
    "level": level
})

print("\n===== AI PROJECT MENTOR =====")
print(ideas)
```

```python
def ai_project_mentor():

    print("===================================")
    print("       🤖 AI PROJECT MENTOR")
    print("===================================")

    choice = input(
        "\nDo you already have a project idea? (yes/no): "
    ).lower()

    if choice == "no":

        domain = input("Enter your preferred domain: ")
        level = input("Enter your skill level: ")

        print("\nGenerating project ideas...\n")

        ideas = idea_chain.invoke({
            "domain": domain,
            "level": level
        })

        print("===== PROJECT IDEAS =====")
        print(ideas)

    elif choice == "yes":

        idea = input("Enter your project idea: ")
        level = input("Enter your skill level: ")
        technology = input("Enter your preferred technology: ")

        print("\nCreating your project plan...\n")

        plan = mentor_chain.invoke({
            "idea": idea,
            "level": level,
            "technology": technology
        })

        print("===== PROJECT PLAN =====")
        print(plan)

    else:
        print("Please enter yes or no.")
```

```python
ai_project_mentor()
```

```python
from gtts import gTTS
from IPython.display import Audio, display

text = str(structure)

# Sentences you DON'T want TTS to read
remove_text = [
    "You are an AI Software Project Mentor.",
    "Design a clean project folder structure.",
    "Include:",
    "1. Main project folder",
    "2. Dataset folder",
    "3. Source code folder",
    "4. Model folder",
    "5. Testing folder",
    "6. Application file",
    "7. Requirements file",
    "8. README file"
]

# Remove only those sentences
for sentence in remove_text:
    text = text.replace(sentence, "")

# Create voice from the remaining text
tts = gTTS(text=text, lang="en")
tts.save("mentor_voice.mp3")

display(Audio("mentor_voice.mp3", autoplay=True))
```
